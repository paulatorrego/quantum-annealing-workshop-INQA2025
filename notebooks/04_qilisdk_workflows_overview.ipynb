# QiliSDK workflows overview

QiliSDK in the workshop is presented as a unified abstraction layer for analog, digital and hybrid workflows.

## Analog module
Hamiltonians, schedules, and time evolution.

```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 10, 200)
driver = 1 - t/10
problem = t/10

plt.figure(figsize=(6,4))
plt.plot(t, driver, label="driver coefficient")
plt.plot(t, problem, label="problem coefficient")
plt.xlabel("time")
plt.ylabel("schedule coefficient")
plt.title("Linear analog schedule")
plt.legend()
plt.show()
```

## Core and Digital modules
Variables, models, tensors, circuits, ansätze, and variational workflows.

```python
best = None
for x in [0,1]:
    for y in [0,1]:
        value = x + 2*y + 2*x*y
        if best is None or value < best[0]:
            best = (value, x, y)
print("Best objective value and assignment:", best)
```
