# Fluxonium with scqubits – gap search

```python
import numpy as np
import matplotlib.pyplot as plt
import scqubits as scq

EJ = 5.0
EC = 0.5
CUTOFF = 140

def make_fluxonium(EL, flux=0.5, cutoff=CUTOFF):
    return scq.Fluxonium(EJ=EJ, EC=EC, EL=EL, flux=flux, cutoff=cutoff)

def gap_for_EL(EL):
    flx = make_fluxonium(EL=EL, flux=0.5)
    evals = flx.eigenvals(evals_count=3)
    return evals[1] - evals[0]
```

```python
EL_grid = np.linspace(0.1, 2.5, 120)
gaps = np.array([gap_for_EL(EL) for EL in EL_grid])
target = 0.5
idx = np.argmin(np.abs(gaps-target))
best_EL = EL_grid[idx]
best_gap = gaps[idx]
print("Best EL:", best_EL)
print("Best gap:", best_gap)

plt.figure(figsize=(6,4))
plt.plot(EL_grid, gaps)
plt.axhline(target, linestyle="--")
plt.axvline(best_EL, linestyle=":")
plt.xlabel("EL [GHz]")
plt.ylabel("Gap E1-E0 [GHz]")
plt.title("Gap tuning at the sweet spot")
plt.show()
```

```python
fluxonium = make_fluxonium(best_EL, flux=0.5)
flux_vals = np.linspace(0.45, 0.55, 81)

evals = []
for fl in flux_vals:
    fluxonium.flux = fl
    vals = fluxonium.eigenvals(evals_count=6)
    vals = vals - vals[0]
    evals.append(vals)
evals = np.array(evals)

plt.figure(figsize=(7,4.5))
for level in range(6):
    plt.plot(flux_vals, evals[:, level], label=f"Level {level}")
plt.xlabel("External flux / Phi0")
plt.ylabel("Energy [GHz], ground subtracted")
plt.title("Low-lying fluxonium spectrum")
plt.legend(ncol=2, fontsize=8)
plt.show()
```
