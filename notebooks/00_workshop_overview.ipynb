# Quantum Hardware Workshop – overview

These notes summarize the workshop in a clean way.

## Main ideas

- Superconducting circuits as a platform for quantum computing and simulation.
- Key elements: capacitors, inductors, Josephson junctions.
- From LC oscillator to qubits through anharmonicity.
- Fluxonium and dispersive readout.
- Practical modeling with `scqubits`.
- QiliSDK as a unified layer for analog, digital and hybrid workflows.


## Workshop practice

The hands-on part focused on:
1. Building a fluxonium model.
2. Finding a parameter `EL` giving about a 500 MHz gap at the sweet spot.
3. Plotting the low-lying spectrum versus flux.
4. Studying coupling to a resonator.
