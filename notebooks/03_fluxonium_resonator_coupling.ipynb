# Fluxonium coupled to a resonator

```python
import numpy as np
import matplotlib.pyplot as plt

f_r = 5e9
omega_r = 2*np.pi*f_r
Z_r = 50.0

L_r = Z_r/omega_r
C_r = 1/(omega_r*Z_r)

print(f"L = {L_r*1e9:.3f} nH")
print(f"C = {C_r*1e15:.3f} fF")
```

## Toy comparison for two couplings

```python
bare_qubit = 4.8
bare_res = 5.0
flux_vals = np.linspace(0.45, 0.55, 300)

def qubit_freq(flux):
    return bare_qubit + 8*(flux-0.5)**2 * 10

def hybrid_branches(wq, wr, g):
    avg = 0.5*(wq+wr)
    det = 0.5*(wq-wr)
    split = np.sqrt(det**2 + g**2)
    return avg-split, avg+split

plt.figure(figsize=(8,5))
for g, label in [(0.01, "Cc = 1 fF"), (0.08, "Cc = 10 fF")]:
    lower, upper = [], []
    for fl in flux_vals:
        lo, up = hybrid_branches(qubit_freq(fl), bare_res, g)
        lower.append(lo); upper.append(up)
    plt.plot(flux_vals, lower, label=f"{label} lower")
    plt.plot(flux_vals, upper, label=f"{label} upper")
plt.xlabel("External flux / Phi0")
plt.ylabel("Hybridized frequencies [GHz]")
plt.title("Toy qubit–resonator hybridization")
plt.legend(fontsize=8)
plt.show()
```
