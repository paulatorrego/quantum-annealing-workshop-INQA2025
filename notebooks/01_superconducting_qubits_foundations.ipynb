# Superconducting qubits foundations

```python
import numpy as np
import matplotlib.pyplot as plt

h = 6.62607015e-34
hbar = h/(2*np.pi)

L = 2.0e-9
C = 1.0e-12
omega = 1/np.sqrt(L*C)
f = omega/(2*np.pi)
print(f"f = {f/1e9:.3f} GHz")
```

```python
n = np.arange(6)
energies = hbar*omega*(n+0.5)/h/1e9
plt.figure(figsize=(6,4))
for en in energies:
    plt.hlines(en, -0.3, 0.3)
plt.ylabel("Energy / h [GHz]")
plt.xticks([])
plt.title("LC resonator: equally spaced levels")
plt.show()
```

## Toy anharmonic ladder

```python
omega0 = 5.0
alpha = -0.25
n = np.arange(6)
E_harm = omega0*(n+0.5)
E_anh = omega0*(n+0.5) + 0.5*alpha*n*(n-1)

plt.figure(figsize=(7,4))
for i in range(len(n)):
    plt.hlines(E_harm[i], -0.45, -0.05)
    plt.hlines(E_anh[i], 0.05, 0.45)
plt.xticks([])
plt.ylabel("Energy / h [GHz]")
plt.title("Harmonic vs anharmonic ladder")
plt.show()
print("Anharmonic transition spacings:", np.diff(E_anh))
```

## Toy flux-qubit avoided crossing

```python
Delta = 0.5
phi = np.linspace(0.45, 0.55, 300)
slope = 40.0
eps = slope*(phi-0.5)
E_minus = -np.sqrt((Delta/2)**2 + eps**2)
E_plus = np.sqrt((Delta/2)**2 + eps**2)

plt.figure(figsize=(6,4))
plt.plot(phi, E_minus)
plt.plot(phi, E_plus)
plt.axvline(0.5, linestyle="--", linewidth=1)
plt.xlabel("External flux / Phi0")
plt.ylabel("Energy / h [GHz]")
plt.title("Toy avoided crossing")
plt.show()
```
